### **Chat word correcter**

```python
chat_words = {'IMHO':'In My Honest/Humble Opinion', 'GN':'Good night','GM':'Good Morning'}
```

```python
def actual(sentence):

  chat_words = {'IMHO':'In My Honest/Humble Opinion', 'GN':'Good night','GM':'Good Morning'}

  for i in sentence:
    if i in string.punctuation:
      sentence = sentence.replace(i,'')


  for i in sentence.split():
    if i.upper() in chat_words:
      sentence = sentence.replace(i,chat_words[i.upper()])

  return sentence
```

```python
actual('gm how are you')
```

**Remove emoji**

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie.😘😘 It was 🔥")
```

# Tokens & Embeddings for LLM Core Architecture: The Transformer








## Tokens :-
>Tokenization ka matlab hai kisi bade sentence ya document ko chhote-chhote tukdon (units) mein todna. In tukdon ko hum Tokens kehte hain.

### 1. Type of Tokenization :-

1.  Word-level Tokenization:-

Har word ek token hota hai.

    Sentence: "AI is amazing"

    Tokens: ["AI", "is", "amazing"]

    Problem: Agar "amazing" ki jagah "amazingly" aa jaye, to AI confuse ho sakta hai agar usne wo word pehle nahi dekha. Isse vocabulary bahut badi ho jati hai.

2. Character-level Tokenization :-

Har letter ek token hota hai.

    Sentence: "AI"

    Tokens: ["A", "I"]

    Problem: Isse sentence bahut lamba ho jata hai aur model ko "matlab" (meaning) nikalne mein mushkil hoti hai.

3. Subword Tokenization :-

(Jo modern AI use karta hai - e.g., GPT, BERT)
Ye words ko aise todta hai ki common hisse (prefixes/suffixes) alag ho jayein.

    Example Word: "Unbelievable"

    Tokens: ["Un", "believ", "able"]

    Benefit: Isse AI naye words ka matlab bhi samajh sakta hai unke tukdon ko dekh kar.

## 2. Embeddings


---



### 1. The Core Concept: Semantic Space
Imagine kijiye ek bada khali kamra (Space). Hum har word ko us kamre mein ek specific jagah (coordinate) par rakhte hain.

    Jo words milte-julte hain, wo paas-paas honge.

    Jo words alag hain, wo door-door honge.

Example:

    "King" aur "Queen" ke coordinates ek doosre ke bahut kareeb honge kyunki dono "Royalty" hain.

    "Apple" aur "Orange" paas honge kyunki dono "Fruits" hain.

    Lekin "King" aur "Apple" ke beech bahut doori hogi.

### 2. High-Dimensional Vectors (Hazaaron Numbers)
Real life mein hum sirf 2D ya 3D space ki baat karte hain, lekin LLMs (jaise GPT-4 ya BERT) 768 ya 1536 dimensions use karte hain.

```python
from transformers import AutoTokenizer, AutoModel
import torch
```

```python
model_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
word = 'apple'

inputs = tokenizer(word,return_tensors='pt')

print(inputs)
```

```python
with torch.no_grad():
  outputs = model(**inputs)
  embeddings = outputs.last_hidden_state
```

### 3. Why do we need this?
    Context (Sahi Matlab): Purane tareekon (One-hot encoding) mein har word independent tha.
    Embeddings se AI ko pata chalta hai ki "Tiger" aur "Cat" mein rishta hai.

    Efficiency: Hazaron words ko 768 numbers mein compress kar dena model ko fast banata hai.

    Generalization: Agar model ne "Dog" ke baare mein kuch seekha hai,
    to wo automatically use "Puppy" par bhi apply kar sakta hai kyunki unke vectors paas-paas hain.

```python
import torch
from transformers import AutoTokenizer, AutoModel
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA
```

```python
model_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
words = ['king','queen','prince','apple','orange','banana','atom']

word_embeddings = []
for word in words:
    inputs = tokenizer(word, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**inputs)
        embeddings = outputs.last_hidden_state[0,1,:].numpy()
        word_embeddings.append(embeddings)

pca = PCA(n_components=2)
reduced_embeddings = pca.fit_transform(np.array(word_embeddings))

plt.figure(figsize=(10,3))

for i, word in enumerate(words):
    plt.scatter(reduced_embeddings[i, 0], reduced_embeddings[i, 1])
    plt.annotate(word, (reduced_embeddings[i, 0] + 0.1, reduced_embeddings[i, 1] + 0.1), fontsize=12)
```

## **Attention Mechanism**



---





**"Ek sentence mein har word ko doosre words par kitna dhyan (attention) dena chahiye?"**

### 1. The Intuition: Query, Key, and Value
Ise ek Library ke example se samjhein:

Query ($Q$): Aap library mein jaakar jo book dhoondh rahe hain (Specific word jise hum process kar rahe hain).

Key ($K$): Books ki spine par likha hua title (Sentence ke baaki sabhi words ka description).

Value ($V$): Book ke andar ki information (Actual content jo hum extract karenge agar Query aur Key match ho jayein).

image.png

```python
import torch
import torch.nn.functional as F

def basic_attention(q, k, v):
    # Step 1: Dot Product (Similarity)
    scores = torch.matmul(q, k.transpose(-2, -1))

    # Step 2: Scaling
    d_k = q.size(-1)
    scaled_scores = scores / torch.sqrt(torch.tensor(d_k, dtype=torch.float32))

    # Step 3: Softmax (Attention Weights)
    weights = F.softmax(scaled_scores, dim=-1)

    # Step 4: Final Output
    output = torch.matmul(weights, v)
    return output, weights

# Example: 1 word, embedding size 4
q = torch.rand(1, 4)
k = torch.rand(1, 4)
v = torch.rand(1, 4)

output, weights = basic_attention(q, k, v)
print("Attention Weights:", weights)
```
