# RAG (Retrieval-Augmented Generation)


---


* RAG is a "Game Changer" concept in the world of AI. In simple terms, it's like letting the AI take an **"Open Book Exam"**.

* Everything we've learned so far (Pre-training) is the AI's "Memory". But if you ask the model something it never studied, it starts making things up (this is called **Hallucination**). RAG is the solution to this problem.


## 1. Concept: What is RAG? (The Librarian Analogy)
Imagine a great Professor (the LLM) who knows almost everything, but hasn't read any news after 2023.

* If you ask him "Who won yesterday's match?", he will guess based on old knowledge (Hallucination).

* In **RAG**, we place an Assistant (the Retriever) next to the Professor who has access to all the latest files and the internet.

* When you ask a question, the Assistant first finds the right file and shows it to the Professor. The Professor then reads that file and gives you the correct answer.


## 2. The 3 Main Steps of RAG (The Technical Flow)
1. **Retrieval (Finding):** Documents or data matching the user's question are pulled from your private database (PDFs, Emails, Company Docs).

2. **Augmentation (Attaching):** That retrieved data is "attached" to the user's original prompt.

3. **Generation (Answering):** The LLM now reads that extra data and generates an accurate answer based on it.


## 3. Practical Implementation:-

```python
from transformers import pipeline

genrator = pipeline('text-generation', model = 'gpt2')

private_knowledge = "In 2026, the World AI Summit was held in Lucknow, India, and the chief guest was a robot named 'Gyan'."


user_query = "Who was the chief guest at the 2026 World AI Summit?"
```

```python
res_simple= genrator(user_query,max_new_tokens = 30)
print(res_simple[0]['generated_text'])
```

```python
rag_prompt = f"Context: {private_knowledge}\n\nQuestion: {user_query}\nAnswer according to the context:"

res_simple= genrator(rag_prompt,max_new_tokens = 30)
print(res_simple[0]['generated_text'])
```

## 4. Key Points (The Benefits)
* **No Hallucinations:** The AI is no longer guessing — it answers from the information it was given.

* **Up-to-date Info:** You don't need to re-train (Fine-tune) the model again and again. Just update the database, and the AI "learns" the new information.

* **Privacy:** You are not handing over your sensitive data (bank records or medical files) for training — you only show it "when needed".


## 5. Lesson Notes: RAG vs. Fine-tuning

* **Fine-tuning** = Teaching a student for 5 years to become a doctor (Hard & Expensive).

* **RAG** = Handing that student a "Medical Textbook" during the exam (Fast & Reliable).


# Vector Databases — RAG (Retrieval-Augmented Generation)


---


* Vector Databases are the "Brain" — or the "Search Engine" — of RAG. If RAG is a Library, then the Vector Database is the Index of that library, which takes you straight to the right book instantly.


To understand this, we first need to understand **Embeddings**.

## 1. Concept: Embeddings and Vector Space
AI doesn't understand words — it understands **Numbers**.
* When we convert a sentence into a long list of numbers, that is called an **Embedding**.
* We can plot these numbers like "Coordinates" ($X, Y, Z$) on a graph.
* **Logic:** Sentences with similar **Meaning** end up **Close** to each other on the graph.

**Example:** The vectors for "King" and "Queen" will be close together, but the vector for "Apple" will be far away from them.


## 2. What is a Vector Database?
It is a special database that stores these "Numbers" (Vectors). When you ask a question:

* Your question is converted into a Vector.

* The database checks which document vector is **Closest** to your question's vector (**Similarity Search**).

* It pulls out the most similar document and hands it to the AI.


## 3. Practical Implementation :-
We don't need to set up a complex database — we can build a small Vector DB using the `sentence-transformers` and `faiss` (Facebook AI Similarity Search) libraries.


### Step A: Install Libraries

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np


model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
documents = [
    "The capital of India is New Delhi.",
    "Python is a popular programming language for Data Science.",
    "The 2026 AI Summit was held in Lucknow.",
    "Almonds are healthy for the brain."
]
```

```python
doc_embeddings = model.encode(documents)
```

```python
doc_embeddings
```

```python
dimension = doc_embeddings.shape
dimension
```

```python
dimension = doc_embeddings.shape[1] # Size of each vector
index = faiss.IndexFlatL2(dimension) # L2 Distance (used to check similarity)
index.add(np.array(doc_embeddings))
```

```python
print(index.ntotal)
```

### Step C: Search (Retrieval)

```python
query = "Where was the AI event held?"


query_embedding = model.encode([query])

d,i = index.search(np.array(query_embedding), k = 1)

print(query)
print(documents[i[0][0]])
```
